# Exercise 17 (6 points)

A pandas DataFrame with hundreds of millions of rows contains an integer identifier column stored as int64, a low-cardinality country column stored as text objects, and a floating-point score column stored as float64. Describe how the dtypes of these three columns could be changed to reduce memory usage and explain the reasoning and any tradeoffs for each change. Please be specific.

For the integer identifier, you could switch from int64 to a smaller type such as int32, which halves the memory used per value, but only if you first verify that every identifier fits within int32's smaller numeric range. For the low-cardinality country column, converting from object to category replaces repeated text strings with compact integer codes, dramatically reducing memory, though it becomes less convenient if many new categories are introduced later. For the floating-point score, using float32 instead of float64 halves its memory footprint, but only if the reduced numerical precision is acceptable for the analysis. The key point is that you should not automatically downcast every column, you must first check whether the smaller data types can safely represent the actual values.

Pandas aligns two series by their index labels rather than by position because the labeles indentify which values belong together, so the values are matched correctly even when the rows appear in different orders. When an index label exsits in one Series but not the other, pandas produces NaN for that label in the result. This label alignment is useful when joining data by an identifier because it prevents values from being combined with the wronf records. For example, if Series A has cattle ID 101 with a price of 150 and Series B has cattle ID 101 with a value of 20, pandas adds those two together by matching on the ID, regardless of where each row sits in its Series.

# Exercise 20 (6 points)

One analyst creates a selection using basic NumPy slicing; another creates a selection using Boolean indexing. Each modifies elements of the resulting selection. Explain whether these changes affect the original array, describe how to ensure independent work, and identify the memory tradeoff. Please be specific.

Basic NumPy slicing creates a view that shares memory with the original array, so modifying the slice also changes the original, while Boolean indexing creates a copy, so modifying it leaves the original untouched. To ensure independent work, call .copy() explicitly—this uses additional memory but prevents unintended changes to the original.

# Exercise 23 (6 points)

Explain why pure functions and immutability (core ideas of functional programming) make unit testing and parallelization easier in a data science pipeline. Give one concrete example of a data transformation that is a good candidate for a pure function rather than a stateful class.

Pure functions always produce the same output for the same input and don't modify external data, while immutability prevents data from being changed in place—together these eliminate hidden side effects, making unit tests predictable and allowing independent operations to run in parallel without interfering with each other. A concrete example is a function that converts cattle weights from pounds to kilograms and returns a new dataset, rather than a stateful class that mutates the original weights in place. Because the conversion function depends only on its input and leaves the original data untouched, it is easy to test with known values and safe to run across multiple chunks of data simultaneously.
