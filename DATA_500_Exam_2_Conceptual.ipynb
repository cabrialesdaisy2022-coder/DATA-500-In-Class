# Exercise 4 (4 points)

Describe what each of the four blocks `try`, `except`, `else`, and `finally` is for, and explain why the chapter recommends putting success-path code in `else` rather than at the end of the `try` block. Please be specific.

try: Contains the code that might raise an exception. It is the "risky" section.

except: Contains the code that handles the exception if one is raised in the try block.

else: Contains code that should run only if the try block completes without raising any exceptions.

finally: Contains code that runs regardless of whether an exception occurred or not (e.g., cleanup actions like closing files).

We use else because putting success-path code there prevents the except block from accidentally catching exceptions that were not caused by the original risky code in the try block. It isolates the error handling to only the specific code we intended to guard.

# Exercise 6 (5 points)

Explain the difference between a bare `raise` and `raise ... from e` inside an `except` block, and explain why the chapter recommends giving each module a single base custom exception (e.g., `TrendlineError`) from which its specific errors inherit. Please be specific.

Bare raise re-raises the current exception while preserving its original traceback. It is used when you want to log or perform a cleanup action but let the original error propagate up the call stack unchanged.

Raise NewError(...) from original_error raises a new exception and explicitly sets the original exception as its __cause__. This creates an explicit chain in the traceback, showing the direct causal link between the two errors (e.g., "The following exception occurred directly because of this one"). This is useful for wrapping low-level errors in higher-level, domain-specific exceptions without losing the original context.

The chapter also recommends having one base custom exception, such as TrendlineError, because it makes errors easier to organize and handle. Specific exceptions can inherit from the base exception, allowing someone to catch either a particular error or all errors related to that module. This gives the programmer more control over how different errors are handled

# Exercise 13 (7 points)

Machine learning code cannot usually be tested with exact-output assertions. Describe the shift in testing mindset, and explain **data leakage** and **training/serving skew**: why no metric reports them and how you would write tests that catch each. Also give two ways to make stochastic tests assertable. Please be specific.

Machine learning models can give slightly different results, so instead of testing for one exact answer, we should check whether the results stay within an acceptable range or meet certain expectations.

Data leakage happens when information that should not be available during training gets included in the model. To prevent this, I would test that the training and testing data are separated and that the model does not use information from the test data during training.

Training/serving skew happens when the data used to train a model is different from the data it receives when making predictions. I would test that the same features, data types, and preprocessing steps are used for both training and prediction.

These problems might not show up in performance metrics because the metrics only measure results on the data being evaluated. They do not automatically check whether the data was leaked or whether the training and prediction processes match.

Two ways to make random tests more reliable are setting a fixed random seed and checking that results fall within an acceptable range instead of expecting an exact number.

# Exercise 14 (7 points)

Define **cohesion** and **coupling**, and explain why good design aims for *high cohesion and low coupling*. Using a function that processes data and also writes to a hard-coded log file, explain how **dependency injection** loosens the coupling and improves testability. Please be specific.

Cohesion describes how closely related the tasks within a function are. High cohesion means a function focuses on one main job instead of trying to do several unrelated things. Coupling describes how much different parts of a program depend on each other. Low coupling means we can change one part without affecting everything else.

Good code should have high cohesion and low coupling because it makes the program easier to understand, maintain, and test.

For example, if a function processes data and automatically saves it to a specific log file, it depends on that file being available. With dependency injection, we can pass in a file or logger instead of hard-coding it inside the function. This allows us to use a fake logger or temporary file when testing, without changing the main function. It makes the code more flexible and easier to test.

# Exercise 17 (4 points)

DRY - Don't Repeat Yourself. This encourages us to avoid repeating the same code by using reusable functions

YAGNI - You Aren't Gonna Need It. This reminds us to not add extra features or make code more complicated than necessary.

If we follow DRY too strictly, we might create complicated functions that are difficult to understand.
