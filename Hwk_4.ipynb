### Exercise 2(a) (12 points)

Create a class called `DataProcessing`. The class should include the following methods:

- `__init__(self, data: pd.DataFrame)`: Initializes the class with a pandas DataFrame.
- `encode_categorical(self, columns)`: Encodes categorical columns using `get_dummies` encoding from `pandas`. 
- `volume(self)`: Computes the volume of the diamond as `x * y * z`.
- `ratio(self)`: Computes the ratio of `depth` and `table`.
- `stand(self)`: Computes the z-score standardization of `carat`. That is, `(carat - mean(carat)) / std(carat)`, where `mean` and `std` are the mean and standard deviation, respectively.
- `fe(self)`: Append the `volume` and `ratio` engineered features to the data frame. Also apply `stand`.

Implement the `DataProcessing` class with the above methods and demonstrate its usage with a sample DataFrame.


```python
class DataProcessing:
    def __init__(self, data: pd.DataFrame):
        self.data = data.copy()

    def encode_catergorical(self, columns):
        self.data = pd.get_dummies(
            self.data, columns=columns, dtype=int
        )
        return self.data 

    def volume(self):
        return self.data["x"] * self.data["y"] * self.data["z"] 

    def ratio(self):
        return self.data["depth"] / self.data["table"]

    def stand(self):
        carat = self.data["carat"]
        self.data["carat"] = (carat - carat.mean()) / carat.std() 

    def fe(self):
        self.data["volume"] = self.volume()
        self.data["depth/table"] = self.ratio()
        self.stand()
        return self.data

# Demonstrate the class with a sample DataFrame
sample = pd.DataFrame({
    "carat": [0.5, 1.0, 1.5],
    "cut": ["Good", "Ideal", "Premium"],
    "depth": [61.0, 62.0, 63.0],
    "table": [55.0, 56.0, 57.0],
    "x": [4.0, 5.0, 6.0],
    "y": [4.1, 5.1, 6.1],
    "z": [2.5, 3.0, 3.5]
}) 

sample_processor = DataProcessing(sample)
sample_processor.encode_catergorical(["cut"])
sample_processor.fe() 
```

In this exercise, we will practice inheritance and polymorphism by extending the `DataProcessing` class built in Exercise 2.

### Exercise 4(a) (4 points)

Create a class called `PriceFlagging` that inherits from `DataProcessing`. The class should include a method `flag_price(self, threshold)` that creates a new column called `price_flag`, using `apply` and a lambda function, that stores 1 if `price` is greater than `threshold` and 0 otherwise.

```python
class PriceFlagging(DataProcessing):
    def flag_price(self, threshold):
        self.data["price_flag"] = self.data["price"].apply(
            lambda price: 1 if price > threshold else 0
        ) 
        return self.data 
```

### Exercise 4(b) (4 points)

Read the `diamonds.csv` data file. Initialize the `PriceFlagging` class, run the inherited `fe` method, and then call `flag_price` with a `threshold` of 10000. Display the first five rows of the resulting data frame.

```python
diamonds = pd.read_csv("diamonds.csv")

price_processor = PriceFlagging(diamonds)
price_processor.fe()
price_processor.flag_price(10000)

price_processor.data.head() 
```
