```
numbers = [5, 12, 3, 8, 7, 9, 10, 4, 15]
```

````python
numbers = [5, 12, 3, 8, 7, 9, 10, 4, 15]

odd_numbers = [num for num in numbers if num % 2 != 0]

average = sum(odd_numbers) / len(odd_numbers) 

print(odd_numbers) 
print(average)
````

# Exercise 7 (8 points)

Chapter 2 introduces two ways to compute the mode (the most frequent value) of a list of integers: a pure-Python loop (`mode_using_loop`) and an implementation that delegates the counting step to `collections.Counter` (`mode_using_counter`).

- Implement both functions exactly as described in Chapter 2.
- Using `numpy`, generate an array of 100,000 random integers between 1 and 1,000 (inclusive) called `random_integers`.
- Use the `timeit` module to time each function over 5 repetitions (`number=5`), storing the results in `loop_time` and `counter_time`, respectively.
- Print both timing results, along with a statement indicating which implementation was faster.

*Note: exact timings will vary depending on your machine, but `mode_using_counter` should be faster.*

````python
import numpy as np
import timeit
from collections import Counter

# Pure-Python loop implementation
def mode_using_loop(data):
    counts = {}
    for value in data:
        if value in counts:
            counts[value] += 1
        else:
            counts[value] = 1
    return max(counts, key=counts.get) 

# Counter implementation
def mode_using_counter(data):
    counts = Counter(data) 
    return counts.most_common(1)[0][0] 

# Generate 100,000 random integers from 1 to 1,000
random_integers = np.random.randint(1, 1001, size=100000) 

# Time each function over 5 repertitions 
loop_time = timeit.timeit(
    lambda: mode_using_loop(random_integers),
    number=5 
)
counter_time = timeit.timeit(
    lambda: mode_using_counter(random_integers),
    number=5 
)  

print("Loop time:", loop_time)
print("Counter time:", counter_time) 

if counter_time < loop_time:
    print("mode_using_counter is faster")
else:
    print("mode_using_loop is faster") 
````
