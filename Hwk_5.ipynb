1. The following four messages are emitted:
"File loaded: 1,000 rows" (INFO)

"12 rows have missing values" (WARNING)

"Column 'age' not found" (ERROR)

"Out of memory" (CRITICAL)

Python's logging level have numerica values:  DEBUG=10, INFO=20, WARNING=30, ERROR=40, CRITICAL=50. Setting level=logging.INFO means the logger will process any message with a level of 20 or higher. Because logging.debug has a value of 10, it is below the threshold and is ignored.

2. Only the WARNING, ERROR, and CRITICAL messages would be emitted. Because if basicConfig is not called, the logging module falls back to a defualt "root" logger. The default threshold for this root logger is WARNING (level 30). Therefore, it will only output messages at level 30 or higher. (The INFO and DEBUG messages would be ignored).

3. Jupyter Notebooks automatically configure the root logger and attach their own handlers so output appears nicely in the browser. The logging.basicConfig() function only works if the root logger has no handlers already attached. Because Jupyter has already set them up, basicConfig silently does nothing. You can fix it by adding force=true to the configuration: logging.basicConfig(level=logging.INFO, force=True). This removes existing handlers and applies your settings.

4. Use logging.exception() only inside an except block. It logs the message at the ERROR level and automatically appends the full traceback of the current exception. This is helpful for debugging exactly where and why a crash happened. If called outside an exception handler, it will log the message as an ERROR, but the traceback part will just say "NoneType: None" because there is no active exception to report.

### Exercise 4(b) (7 points)

Write a function called `column_summary(df, column)` that returns a dictionary with the mean, standard deviation, and median (rounded to 2 decimals) of `column`. The function must:

- Log an `INFO` message when it starts.
- Raise a `ValueError` with an informative message (what was expected, what was received) if `column` is not in `df`.
- Use `try`, `except`, `else`, and `finally` blocks to convert the column to numeric with `pd.to_numeric`. If the column cannot be converted, log the error using `logging.exception` and return a dictionary with the same keys and `np.nan` values (so the return format is consistent). If it can, log an `INFO` message with the summary and return it.
- Log an `INFO` message saying that the processing finished, whether or not an error occurred.

Keep the `try` block as small as possible.

````python
def column_summary(df, column):
    logging.info(f"Starting summary for column: {column}")

    if column not in df.columns:
        raise ValueError(f"Expected column '{column}', but not recieved a column that does not exist.")
    try:
        numeric_data = pd.to_numeric(df[column], errors="raise")
    except (ValueError, TypeError):
        logging.exception(f"Could not convert column '{column}' to numeric")
        return {
            "mean": np.nan,
            "std": np.nan,
            "median": np.nan
        }
    else:
        result = {
            "mean": round(numeric_data.mean(), 2),
            "std": round(numeric_data.std(), 2),
            "median": round(numeric_data.median(), 2)
        }
        logging.info(f"Successfully summarized column: {column}")
        return result
    finally:
        logging.info(f"Finished processing column: {column}")
````
