1. Documetation that is "close to the code" lives directly in the function or class signature, so the reader sees it in the glance as the logic itself. Documentation that is "far from the code" (a README, a wiki page) lives in separate file that is not checked by any tool, so it can silently drift out of sync as the code changes. It is recommended to push information as far up the hierarchy as possible because the layers closer to the code are cheaper to maintain and cannot "silently disagree" with the code.

2. A comment is an inline, lightweight note meant to explain why a specific piece of code does something non-obvious. A docstring is the formal, structed documnetation of a function, class, or module. It systematically describes the function's behavior, parameters, return, value, exceptions, and edge cases.

3. Duplicated documentation is worse than missing documentation because when the same fact is written in two or more places, there is no mechanism keeping those copies synchronized. Sooner or later the code changes and only one copy is updated, and nowthe reader has no way to know which version is correct. Example: a parameter's valid range for an alignment is documented both in the function's docstring ("must be non-negative") and in the project's README. If thr function is later changed to allowed negative threshold, and the docstring is updated, the README now contradicts the actual behavior, and a new user following the README may pass a value that disagrees with what the code now expects.

In this exercise, we will practice documenting code the way Chapter 9 describes: with a descriptive name, type hints, and a complete Google-style docstring (summary, `Args`, `Returns`, `Raises`, and `Example`).

Consider the following undocumented, poorly named function:

```python
def f(df, n):
    df = df[df['price'] >= n]
    df['price'] = df['price'].astype(float)
    return df
```

Your task:

- **Rename the function** to something descriptive that follows the `verb_noun` convention (e.g., a function that filters rows should start with a verb like `filter_`).
- **Add type hints** to the parameters and the return value (the input and output are `pandas.DataFrame`; the threshold is a `float`).
- **Add a `ValueError`** that is raised if the threshold `n` is negative.
- **Write a complete Google-style docstring** including a one-line summary, `Args`, `Returns`, `Raises`, and a runnable `Example`.

Then apply your documented function to the following data, using a minimum price threshold of `20`.

```python
import pandas as pd

data = {'item': ['A', 'B', 'C', 'D'],
        'price': ['10', '25', '30', '15']}
df = pd.DataFrame(data)
```

The expected outcome is shown below.

| item | price |
| --- | --- |
| B | 25.0 |
| C | 30.0 |

````python
import pandas as pd


def filter_by_min_price(df: pd.DataFrame, min_price: float) -> pd.DataFrame:
    """Keep rows with prices at or above the minimum price.

    Args:
        df: A DataFrame with a price column that can be converted to floats.
        min_price: The minimum price to keep, including the threshold itself.

    Returns:
        A filtered DataFrame with prices stored as floats.

    Raises:
        ValueError: If min_price is negative or a price cannot be converted
            to a float.

    Example:
        >>> items = pd.DataFrame({"item": ["A", "B"], "price": ["10", "25"]})
        >>> filter_by_min_price(items, 20.0)["item"].tolist()
        ['B']
    """
    if min_price < 0:
        raise ValueError("The minimum price must be nonnegative.")

    items = df.copy()
    items["price"] = items["price"].astype(float)
    return items.loc[items["price"] >= min_price]


data = {
    "item": ["A", "B", "C", "D"],
    "price": ["10", "25", "30", "15"],
}
df = pd.DataFrame(data)

filter_by_min_price(df, 20.0)
````
